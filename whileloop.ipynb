* Q1. Print numbers from 1 to 10 using a while loop.
* Q2. Print numbers from 10 to 1 using a while loop.
* Q3. Print all even numbers from 1 to 20.
* Q4. Print all odd numbers from 1 to 20.
* Q5. Take a number from the user and print numbers from 1 to that number.

```python
# Q1. Print numbers from 1 to 10 using a while loop.
x=1
while x<=10:
    print(f"{x}")
    x+=1 
```

```python
# Q2. Print numbers from 10 to 1 using a while loop.
x=10
while x>0:
    
    print(x)
    x=x-1 
```

```python
# Q3. Print all even numbers from 1 to 20.
x=1
while x <=20:
    if x % 2==0:
        print(f"Even Numbers: {x}")
    x+=1
```

```python
# Q4. Print all odd numbers from 1 to 20.
x=0
while x<20:
    x+=1
    if x % 2 !=0:
        print(f"Odd Numbers: {x}")
```

```python
# Q5. Take a number from the user and print numbers from 1 to that number.
num=eval(input("Enetr num: "))
x=1
while x<=num:
    print(x)
    x+=1 
```

* Q6. Find the sum of numbers from 1 to 10.
* Q7. Take a number from the user and print its multiplication table from 1 to 10.
* Q8. Take a number from the user and print all its factors.
* Q9. Take a number from the user and check whether it is even or odd.
* Q10. Take a number from the user and count the number of digits in it.

```python
# Q6. Find the sum of numbers from 1 to 10.
total=0
x=1
while x <= 10:
    total=total+x
    x+=1    
print(f"Total number of sum = {total}")
```

```python
# Q7. Take a number from the user and print its multiplication table from 1 to 10.
num=int(input("Enter Number:"))
x=1
while x<=10:
    print(f"{num} X {x} = {num*x}")
    x+=1
print(f"Multiplication Table of: {num}")
```

```python
# Q8. Take a number from the user and print all its factors.
num=int(input("Enter Number:"))
x=1
while x<=num:
    if num % x==0:
        print(x)
    x+=1
print(f"factors of {num} ")
```

```python
# Q9. Take a number from the user and check whether it is even or odd.
num=int(input("Enter Number:"))
x=1
while x<=1:
    if num % 2==0:
        print(f"{num} is Even ")
    else:
        print(f"{num} is Odd ")
    x+=1
```

```python
# Q10. Take a number from the user and count the number of digits in it.
num=int(input("Enter Number:"))

count=0
while num>1:
    count+=1
    num =num // 10
print(count)
```

* Q11. Take a number from the user and find the sum of its digits.
* Q12. Take a number from the user and find its factorial.

```python
# Q11. Take a number from the user and find the sum of its digits.
num = int(input("Enter Number: "))

total = 0

while num > 0:
    digit = num % 10
    total = total + digit
    num = num // 10
print(total)
```

```python
#  Q12. Take a number from the user and find its factorial.
num=int(input("Enter Number:"))
fact =1
x =1
while x <= num:
    fact=fact * x
    x+=1
    
print(fact)
```
